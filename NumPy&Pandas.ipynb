```python
import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

**JAWABAN LATIHAN 1** :

Hasil dari perintah harga * 0.9 pada NumPy array akan otomatis mengalikan setiap elemen angka di dalamnya dengan 0.9 untuk menghitung diskon. Jika hanya menggunakan list Python biasa ([5000, 7000, 3000] * 0.9), eksekusi akan menghasilkan TypeError. Operasi perkalian pada list bawaan Python tidak menerapkan fungsi matematis vectorized ke setiap elemen, melainkan hanya mendukung perkalian dengan bilangan bulat (integer) untuk menggandakan atau mengulang urutan elemen di dalam list tersebut.

```python
import pandas as pd

data_kantin = {
 'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
 'harga': [12000, 4000, 10000, 4000, 8000],
 'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

**JAWABAN LATIHAN 2** :

Berdasarkan pendefinisian data_kantin, kolom menu memiliki data kosong (None) pada baris kelima, dan kolom terjual memiliki data kosong pada baris ketiga. Jika data ini langsung dianalisis tanpa dibersihkan, fungsi agregasi (seperti .sum() atau .mean()) dapat menghasilkan error, perhitungan total pendapatan menjadi tidak valid, dan insight yang dihasilkan akan bias.

```python
import pandas as pd
df = pd.read_csv('data_kantin.csv')

print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

**  JAWABAN LATIHAN 3 **:

Jika hasil dari df.info() menunjukkan ada kolom yang jumlah non-null-nya lebih sedikit dari jumlah baris total dataset, artinya kolom tersebut mengandung missing value (data kosong, NaN, atau None).

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

**JAWABAN LATIHAN 4** :

Metode fillna(0) digunakan pada kolom terjual karena sangat logis secara bisnis jika sebuah menu memang tidak laku atau terjual sebanyak 0 porsi pada periode tersebut. Sebaliknya, baris dengan kolom menu yang kosong dihapus menggunakan dropna karena entitas utamanya tidak diketahui. Jika nama barang tidak ada, data harga dan jumlah terjual kehilangan konteksnya sehingga tidak relevan untuk dipertahankan.

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
```

**JAWABAN LATIHAN 5** :

Setelah menerapkan fungsi drop_duplicates(), jumlah baris dataset menjadi 4. Terdapat satu baris duplikat ('Es Teh' dengan harga 4000) yang dihapus dari total 5 baris awal. Memastikan tipe data sudah benar (seperti integer untuk harga) menggunakan dtypes sangat esensial agar operasi manipulasi lanjutan seperti perkalian antara harga dan jumlah terjual dapat dieksekusi tanpa kendala komputasi.

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

**JAWABAN LATIHAN 6** :

Berdasarkan agregasi dari fungsi groupby, 'Nasi Goreng' menghasilkan total_pendapatan tertinggi (Rp276.000). Informasi ini sangat krusial untuk pengambilan keputusan operasional kantin. Pengelola dapat menggunakan insight ini untuk meningkatkan stok bahan baku Nasi Goreng, menempatkannya sebagai menu unggulan promosi, atau mengevaluasi menu dengan pendapatan terendah.
